```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neural_network import MLPClassifier
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import cross_val_score, cross_validate
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
import xgboost as xgb
```

```python
# Load preprocessed data
# If there is no matching csv, then copy the code in this notebook into the preprocessing notebook
# Or download the dataframe to csv in the preprocessing notebook
data = pd.read_csv('fake_news_data.csv')
```

```python
# Separate features and classes
X = data.drop(columns=['fakeNews', 'subject', 'date']) # To view the importance of 'subject' and 'date', remove them from the list
y = data['fakeNews'] 

# Scale the features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
# Decision Tree to test feature importance
# To view importance of subject and date features, go to cell above and keep them in the dataframe
# Split dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=67)

# Scale
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Initialize and train the tree
dt_model = DecisionTreeClassifier(random_state=42)
dt_model.fit(X_train_scaled, y_train)

# Get feature importances
feature_importance_dt = dt_model.feature_importances_

# Rank features by importance
sorted_idx_dt = np.argsort(feature_importance_dt)[::-1]

# Lists of feature importance and feature names
feature_importance = []
feature_names = []
# Fill the lists
for idx in sorted_idx_dt[:10]: 
  feature_importance.append(feature_importance_dt[idx])
  feature_names.append(X.columns[idx])

# Plot importance with bar plot
barplot = plt.bar(x=list(range(1, 11)), height=feature_importance)
plt.title("Most Important Features")
plt.ylabel("Importance")
plt.xlabel("Features")
plt.xticks(list(range(1, 11)), feature_names)
plt.show()
```

```python
# Neural Network
nn_model = MLPClassifier(hidden_layer_sizes=(100, 50),
                    activation='relu', 
                    max_iter=500,
                    random_state=67)

# Four scoring metrics
scoring = ['accuracy', 'precision', 'f1', 'recall']

# 5-fold cross-validation
nn_cv_scores = cross_validate(nn_model, X_scaled, y, cv=5, n_jobs=-1, scoring=scoring)
print(f'{nn_model.__class__.__name__} - Mean Accuracy: {nn_cv_scores['test_accuracy'].mean():.4f}\nMean Precision: {nn_cv_scores['test_precision'].mean():.4f}\nMean F1: {nn_cv_scores['test_f1'].mean():.4f}\nMean Recall: {nn_cv_scores['test_recall'].mean():.4f}\n')
```

```python
# SVM 
svm_model = SVC(random_state=67)

# Same scoring metrics
svm_cv_scores = cross_validate(svm_model, X_scaled, y, cv=5, n_jobs=-1, scoring=scoring)
print(f'{svm_model.__class__.__name__} - Mean Accuracy: {svm_cv_scores['test_accuracy'].mean():.4f}\nMean Precision: {svm_cv_scores['test_precision'].mean():.4f}\nMean F1: {svm_cv_scores['test_f1'].mean():.4f}\nMean Recall: {svm_cv_scores['test_recall'].mean():.4f}\n')
```

```python
# Logistic Regression
lr_model = LogisticRegression()

lr_cv_scores = cross_validate(lr_model, X_scaled, y, cv=5, n_jobs=-1, scoring=scoring)
print(f'{lr_model.__class__.__name__} - Mean Accuracy: {lr_cv_scores['test_accuracy'].mean():.4f}\nMean Precision: {lr_cv_scores['test_precision'].mean():.4f}\nMean F1: {lr_cv_scores['test_f1'].mean():.4f}\nMean Recall: {lr_cv_scores['test_recall'].mean():.4f}\n')
```

```python
# Random Forest
rf_model = RandomForestClassifier(n_estimators=100, random_state=67)

rf_cv_scores = cross_validate(rf_model, X_scaled, y, cv=5, n_jobs=-1, scoring=scoring)
print(f'{rf_model.__class__.__name__} - Mean Accuracy: {rf_cv_scores['test_accuracy'].mean():.4f}\nMean Precision: {rf_cv_scores['test_precision'].mean():.4f}\nMean F1: {rf_cv_scores['test_f1'].mean():.4f}\nMean Recall: {rf_cv_scores['test_recall'].mean():.4f}\n')
```

```python
# XGBoost
xgb_model = xgb.XGBClassifier()

xgb_cv_scores = cross_validate(xgb_model, X_scaled, y, cv=5, n_jobs=-1, scoring=scoring)
print(f'{xgb_model.__class__.__name__} - Mean Accuracy: {xgb_cv_scores['test_accuracy'].mean():.4f}\nMean Precision: {xgb_cv_scores['test_precision'].mean():.4f}\nMean F1: {xgb_cv_scores['test_f1'].mean():.4f}\nMean Recall: {xgb_cv_scores['test_recall'].mean():.4f}\n')
```
